```python
from moviepy.editor import VideoFileClip
from pathlib import Path

# Folder containing the videos
folder = Path('/Volumes/EntwPsy/Studies/ChildLens/videos/annotated_superannotate')

# File extensions to include (adjust as needed)
video_extensions = ['.mp4', '.MP4', '.mov', '.MOV', '.mkv', '.MKV']

total_seconds = 0

# Loop through video files
for video_file in folder.iterdir():
    if video_file.suffix in video_extensions:
        try:
            clip = VideoFileClip(str(video_file))
            duration = clip.duration  # duration in seconds
            total_seconds += duration
            print(f"{video_file.name}: {int(duration // 3600):02}:{int((duration % 3600) // 60):02}:{int(duration % 60):02}")
            clip.close()
        except Exception as e:
            print(f"Could not process {video_file.name}: {e}")

# Convert total seconds to hh:mm:ss
hours = int(total_seconds // 3600)
minutes = int((total_seconds % 3600) // 60)
seconds = int(total_seconds % 60)

print(f"\nTotal duration: {hours:02}:{minutes:02}:{seconds:02}")
```
